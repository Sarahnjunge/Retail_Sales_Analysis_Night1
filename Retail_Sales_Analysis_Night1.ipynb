```python

import pandas as pd

data = {
    "date": [
        "2026-10-20", "2026-10-20", "2026-10-20",
        "2026-10-21", "2026-10-21", "2026-10-21",
        "2026-10-22", "2026-10-22", "2026-10-22",
        "2026-10-23", "2026-10-23", "2026-10-23"
    ],
    "product": [
        "Bread", "Milk", "Eggs",
        "Bread", "Milk", "Rice",
        "Eggs", "Bread", "Rice",
        "Milk", "Eggs", "Rice"
    ],
    "units_sold": [10, 5, 3, 8, 7, 2, 6, 12, 4, 9, 4, 3],
    "unit_price": [60, 70, 450, 60, 70, 250, 450, 60, 250, 70, 450, 250],
    "unit_cost": [40, 50, 380, 40, 50, 210, 380, 40, 210, 50, 380, 210]
}

df = pd.DataFrame(data)
df["date"] = pd.to_datetime(df["date"])

print("Dataset created successfully!")
print("Shape:", df.shape)
df.head()
```

```python
df["revenue"] = df["units_sold"] * df["unit_price"]
df["cost"] = df["units_sold"] * df["unit_cost"]
df["gross_profit"] = df["revenue"] - df["cost"]
df.head()
```

```python
df.groupby("product")["revenue"].sum()
```

```python
df.groupby("product")["gross_profit"].sum()
```

```python
import matplotlib.pyplot as plt

revenue_by_product = df.groupby("product")["revenue"].sum()

revenue_by_product.plot(kind="bar")

plt.title("Total Revenue by Product")
plt.xlabel("Product")
plt.ylabel("Revenue (KSh)")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

### Business Insight: Revenue by Product

Eggs generated the highest total revenue at KSh 5,850, while milk generated the lowest at KSh 1,470 during the four-day period.

This comparison helps the shop owner understand which products contribute most to sales revenue. However, revenue alone does not determine which products are most profitable. Gross profit, customer demand, and stock availability should also be considered before making inventory decisions.
