```python

import pandas as pd

data = {
    "date": [
        "2026-10-20", "2026-10-20", "2026-10-20",
        "2026-10-21", "2026-10-21", "2026-10-21",
        "2026-10-22", "2026-10-22", "2026-10-22",
        "2026-10-23", "2026-10-23", "2026-10-23"
    ],
    "product": [
        "Bread", "Milk", "Eggs",
        "Bread", "Milk", "Rice",
        "Eggs", "Bread", "Rice",
        "Milk", "Eggs", "Rice"
    ],
    "units_sold": [10, 5, 3, 8, 7, 2, 6, 12, 4, 9, 4, 3],
    "unit_price": [60, 70, 450, 60, 70, 250, 450, 60, 250, 70, 450, 250],
    "unit_cost": [40, 50, 380, 40, 50, 210, 380, 40, 210, 50, 380, 210]
}

df = pd.DataFrame(data)
df["date"] = pd.to_datetime(df["date"])

print("Dataset created successfully!")
print("Shape:", df.shape)
df.head()
```

```python
df["revenue"] = (["unit_price" - "unit_cost"])
```

```python
df.groupby("product")["revenue"].sum()
```

```python
df.groupby("product")["gross_profit"].sum()
```

```python
df.groupby("date")["revenue"].sum()
```
